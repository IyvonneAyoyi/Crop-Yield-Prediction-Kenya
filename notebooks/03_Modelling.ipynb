# Model Training

This notebook trains multiple machine learning regression models to predict crop yield using the preprocessed datasets.

The following models will be trained:

- Linear Regression
- Random Forest Regressor
- Gradient Boosting Regressor
- XGBoost Regressor

The trained models will be saved for later evaluation and deployment.

```python
import pandas as pd
import joblib

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor

from xgboost import XGBRegressor
```

## Load the Preprocessed Datasets

Load the training and testing datasets created during preprocessing. The scaled datasets will be used for Linear Regression, while the unscaled datasets will be used for the tree-based models.

```python
# Load encoded datasets

X_train = pd.read_csv("../Data/Processed_data/X_train.csv")
X_test = pd.read_csv("../Data/Processed_data/X_test.csv")

y_train = pd.read_csv("../Data/Processed_data/y_train.csv").squeeze()
y_test = pd.read_csv("../Data/Processed_data/y_test.csv").squeeze()

# Load scaled datasets

X_train_scaled = pd.read_csv("../Data/Processed_data/X_train_scaled.csv")
X_test_scaled = pd.read_csv("../Data/Processed_data/X_test_scaled.csv")

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("Scaled training set:", X_train_scaled.shape)
print("Scaled testing set:", X_test_scaled.shape)
```

## Train the Linear Regression Model

Linear Regression is used as a baseline model for crop yield prediction. Since Linear Regression is sensitive to differences in feature scales, the scaled training dataset is used to fit the model. The trained model will then be saved for later evaluation and deployment.

```python
# Create the Linear Regression model
linear_model = LinearRegression()

# Train the model
linear_model.fit(X_train_scaled, y_train)

print("Linear Regression model trained successfully!")
```

## Save the Linear Regression Model

The trained Linear Regression model is saved to the `models` directory. Saving the model allows it to be reused later for evaluation and deployment without retraining.

```python
# Save the trained Linear Regression model
joblib.dump(linear_model, "../models/linear_regression_model.pkl")

print("Linear Regression model saved successfully!")
```

## Train the Random Forest Regressor

Random Forest Regressor is an ensemble learning algorithm that combines multiple decision trees to improve prediction accuracy and reduce overfitting. Unlike Linear Regression, Random Forest does not require feature scaling, so the unscaled training dataset is used.

```python
# Create the Random Forest model
random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train the model
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

## Save the Random Forest Model

The trained Random Forest model is saved to the `models` directory so that it can be reused during model evaluation and application deployment.

```python
# Save the trained Random Forest model
joblib.dump(
    random_forest_model,
    "../models/random_forest_model.pkl"
)

print("Random Forest model saved successfully!")
```

## Train the Gradient Boosting Regressor

Gradient Boosting Regressor is an ensemble learning algorithm that builds decision trees sequentially. Each new tree attempts to correct the prediction errors made by the previous trees, often resulting in higher predictive accuracy. Like Random Forest, it uses the unscaled training dataset.

```python
# Create the Gradient Boosting model
gradient_boosting_model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)

# Train the model
gradient_boosting_model.fit(X_train, y_train)

print("Gradient Boosting model trained successfully!")
```

## Save the Gradient Boosting Model

The trained Gradient Boosting Regressor is saved to the `models` directory for later evaluation and deployment.

```python
# Save the trained Gradient Boosting model
joblib.dump(
    gradient_boosting_model,
    "../models/gradient_boosting_model.pkl"
)

print("Gradient Boosting model saved successfully!")
```

## Train the XGBoost Regressor

XGBoost (Extreme Gradient Boosting) is an advanced ensemble learning algorithm that builds decision trees sequentially while optimizing prediction errors. It incorporates regularization techniques to improve generalization and often achieves high predictive performance on structured datasets. The unscaled training dataset is used to train the model.

```python
# Create the XGBoost model
xgboost_model = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=42
)

# Train the model
xgboost_model.fit(X_train, y_train)

print("XGBoost model trained successfully!")
```

## Save the XGBoost Model

The trained XGBoost Regressor is saved to the `models` directory for later evaluation and deployment.

```python
# Save the trained XGBoost model
joblib.dump(
    xgboost_model,
    "../models/xgboost_model.pkl"
)

print("XGBoost model saved successfully!")
```
